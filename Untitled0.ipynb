```python
import torch
print("GPU available:", torch.cuda.is_available())

!pip install transformers datasets scikit-learn matplotlib seaborn

import pandas as pd
import numpy as np
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification, Trainer, TrainingArguments
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_curve, auc
import matplotlib.pyplot as plt
import seaborn as sns
from datasets import load_dataset


```

```python
df = pd.read_csv("/naitik_moviedata.csv")

# Ensure correct column names
df.columns = ["review", "sentiment"]

# Convert labels
df["label"] = df["sentiment"].map({"positive": 1, "negative": 0})

# Reduce dataset for fast training (4000 rows instead of 50k)
df_small = df.sample(2500, random_state=42)

# Train-test split
train_texts, test_texts, train_labels, test_labels = train_test_split(
    df_small["review"].tolist(),
    df_small["label"].tolist(),
    test_size=0.2,
    random_state=42
)

print("Training samples:", len(train_texts))
print("Testing samples:", len(test_texts))
```

```python
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")

train_encodings = tokenizer(train_texts, truncation=True, padding=True, max_length=256)
test_encodings  = tokenizer(test_texts,  truncation=True, padding=True, max_length=256)

class IMDbDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels
    def __getitem__(self, idx):
        return {
            'input_ids': torch.tensor(self.encodings['input_ids'][idx]),
            'attention_mask': torch.tensor(self.encodings['attention_mask'][idx]),
            'labels': torch.tensor(self.labels[idx])
        }
    def __len__(self):
        return len(self.labels)

train_dataset = IMDbDataset(train_encodings, train_labels)
test_dataset  = IMDbDataset(test_encodings, test_labels)
```

```python
# Disable Weights & Biases logging completely
import os
os.environ["WANDB_DISABLED"] = "true"
os.environ["WANDB_MODE"] = "disabled"

from transformers import AutoModelForSequenceClassification, Trainer, TrainingArguments

# Load model
model = AutoModelForSequenceClassification.from_pretrained(
    "distilbert-base-uncased",
    num_labels=2
)

# Training arguments (simple, clean, compatible)
training_args = TrainingArguments(
    output_dir="/content/sentiment_model",
    num_train_epochs=1,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    learning_rate=5e-5,
    logging_steps=200,
     report_to=[]        # disable WANDB/logging services
)

# Metric function
def compute_metrics(pred):
    logits, labels = pred
    preds = np.argmax(logits, axis=1)
    acc = accuracy_score(labels, preds)
    return {"accuracy": acc}

# Trainer setup
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
    compute_metrics=compute_metrics
)

# Train model
trainer.train()

```

```python
raw_preds = trainer.predict(test_dataset)
pred_logits = raw_preds.predictions
pred_labels = np.argmax(pred_logits, axis=1)

accuracy = accuracy_score(test_labels, pred_labels)
print("Accuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(test_labels, pred_labels, target_names=["negative", "positive"]))

```

```python
cm = confusion_matrix(test_labels, pred_labels)

plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["negative","positive"],
            yticklabels=["negative","positive"])
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.savefig("confusion_matrix.png")
plt.show()

```

```python
probs = torch.softmax(torch.tensor(pred_logits), dim=1)[:,1].numpy()

fpr, tpr, thresholds = roc_curve(test_labels, probs)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(6,5))
plt.plot(fpr, tpr, label=f"AUC = {roc_auc:.3f}")
plt.plot([0,1],[0,1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.savefig("roc_curve.png")
plt.show()
```

```python
def predict(text):
    enc = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=256)
    # Move input tensors to the same device as the model
    enc = {k: v.to(model.device) for k, v in enc.items()}
    with torch.no_grad():
        logits = model(**enc).logits
    # Move logits to CPU before converting to numpy
    probs = torch.softmax(logits, dim=1).cpu().numpy()[0]
    label = np.argmax(probs)
    return label, probs[label]

samples = [
    "This movie was absolutely fantastic!",
    "The acting was terrible and the story was pointless.",
    "Not the worst thing ever but definitely boring in parts.",
]

for s in samples:
    label, confidence = predict(s)
    print(f"Text: {s}\nPrediction: {['Negative','Positive'][label]} ({confidence:.3f})\n")
```
